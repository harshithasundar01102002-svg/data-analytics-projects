```python
## The cost column in the dataframe “Project” has some missing values. 
#These missing values replace the missing values by running average. Use the “For” loop for this task. 
```

```python
project = pd.read_csv("project.csv")
```

```python
(project)
```

```python
running_sum = 0
count = 0

# calculate running average and replacing the missing values using the if statement
for i in range(len(project)):
    if pd.notna(project.loc[i, "cost"]):
        running_sum += project.loc[i, "cost"]
        count += 1
    else:
        avg = running_sum / count
        project.loc[i, "cost"] = avg
        running_sum += avg
        count += 1
#display the update of project 
(project)
```

```python
## Add the cost of all projects for each Employee and save it in new dataframe “TotalProjCost” with three columns ID, First Name, and Total cost 
```

```python
#calculate total project cost for each employee
total_cost = project.groupby("id")[["cost"]].sum().reset_index()

#merge with employee DataFrame
total_project_cost = pd.merge(
    employee[["id","First Name"]],
    total_cost,
    on ="id",
    how = "inner"
)

#rename the cost column 
total_project_cost.rename(columns={"cost": "total_cost"}, inplace = True )

#save as csv files
total_project_cost.to_csv("total_project_cost.csv", index = False )

#display the DataFrame
(total_project_cost)
```
